# Esercizio

```python
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
from sklearn.model_selection import cross_val_score # will be used to separate training and test
import numpy as np

import graphviz
```

```python
def evaluate_model(y_test, y_predicted):
    # print some metrics results
    acc_score = accuracy_score(y_test, y_predicted)
    print(f"Accuracy score:\t{acc_score}")
    f1 = f1_score(y_test, y_predicted, average='macro')
    print(f"F1 score:\t{f1}")

def save_and_plot_tree(model, labels, feature_names):
    dot_data = tree.export_graphviz(model, out_file=None)
    graph = graphviz.Source(dot_data)

    dot_data = tree.export_graphviz(model, out_file=None,
                         feature_names=feature_names,
                         class_names=labels,
                         filled=True, rounded=True,
                         special_characters=True)
    graph = graphviz.Source(dot_data)
    return graph

def split_ds(x, y, seed: int=0, test_size: float=0.1):
    assert test_size <= 1 and test_size >= 0.01

    # Generate a random permutation of the indices of examples that will be later used
    # for the training and the test set
    np.random.seed(seed)
    indices = np.random.permutation(len(y))

    # We now decide to keep the last 10 indices for test set, the remaining for the training set
    test_size=int(len(y)*test_size)
    indices_training=indices[:-test_size]
    indices_test=indices[-test_size:]

    X_train = x[indices_training] # keep for training all the matrix elements with the exception of the last 10
    y_train = y[indices_training]
    X_test  = x[indices_test] # keep the last 10 elements for test set
    y_test  = y[indices_test]

    return X_train, X_test, y_train, y_test
```

```python
from sklearn.datasets import load_iris
iris = load_iris()

# Split the dataset
iris_X_train, iris_X_test, iris_y_train, iris_y_test = split_ds(iris.data, iris.target)
```

## TODO 3

Si apprendano gli alberi cercando di evitare l'overfitting (migliorando l'errore sul test set) facendo 'tuning' degli iper-parametri: il minimo numero dei campioni per foglia, la massima profondità dell'albero, i parametri di minomo decremento dell'impurezza, massimo numero dei nodi foglia, ecc. Si consiglia di usare la classe gridSearchCV per questo tipo di tuning di molti parametri.

```python
from sklearn.model_selection import GridSearchCV

clf = tree.DecisionTreeClassifier(random_state=300, class_weight={0:1,1:1,2:1})

params = {
  "criterion": ["gini", "entropy", "log_loss"], # Function to measure the quality of a split.
  "max_depth": range(3, 8),                     # Maximum depth of the tree.
  "min_samples_leaf": range(2, 8),              # Minimum number of samples required to be at a leaf node.
  "max_leaf_nodes": range(3, 10),               # Grow a tree with max_leaf_nodes in best-first fashion.
  "min_samples_split": range(2, 10),            # Minimum number of samples required to split an internal node
  "min_impurity_decrease": np.linspace(0, 2, 11).tolist(),  # A node will be split if this split induces a decrease of the impurity greater than or equal to this value.
}

grid_search = GridSearchCV(clf, params)
grid_search.fit(iris.data, iris.target)
```

```python
grid_search.best_params_
```

```python
predicted_y_test = grid_search.predict(iris_X_test)

# Print f1 score and accuracy of the model
evaluate_model(iris_y_test, predicted_y_test)
```

```python
save_and_plot_tree(grid_search.best_estimator_, iris.target_names, iris.feature_names)
```

## TODO 5

Si costruiscano le curve ROC (o curve nello spazio di coverage) e le si mostri per ciascun modello ad albero creato su un problema binario (con 1 sola classe positiva): per ciascun modello dovete costruire tre curve, una per ciascuna classe, considerata a turno la classe positiva.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay
from sklearn import metrics
```

```python
def train_test_split_bin_clf(X, Y, label):
  label_indices = np.where(Y == label)[0]
  mask = np.full_like(Y, False)
  mask[label_indices] = True

  X_train, X_test, y_train, y_test = split_ds(X, mask, test_size=0.1, seed=42)

  return X_train, y_train, X_test, y_test
```

```python
labels = np.unique(iris.target)
clf = grid_search.best_estimator_

figure, axis = plt.subplots(1, len(labels), figsize=(15, 5),)

for idx, label in enumerate(labels):
    X_train, y_train, X_test, y_test = train_test_split_bin_clf(iris.data, iris.target, label)

    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)

    fpr, tpr, _ = metrics.roc_curve(y_test, y_pred)
    auc = round(metrics.roc_auc_score(y_test, y_pred), 4)
    axis[idx].plot(fpr,tpr,label=f"AUC={auc}")

    axis[idx].legend(loc='lower right')
    axis[idx].set_title(iris.target_names[label])
    axis[idx].grid()

plt.show()
```
