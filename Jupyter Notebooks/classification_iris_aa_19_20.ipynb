# Declare the type of prediction model and the working criteria for the model induction algorithm

```python
clf = tree.DecisionTreeClassifier(criterion="entropy",random_state=300,min_samples_leaf=5,class_weight={0:1,1:1,2:1})
```

# Split the dataset in training and test set

```python
# Generate a random permutation of the indices of examples that will be later used
# for the training and the test set
import numpy as np
np.random.seed(0)
indices = np.random.permutation(len(iris.data))

# We now decide to keep the last 10 indices for test set, the remaining for the training set
indices_training=indices[:-10]
indices_test=indices[-10:]

iris_X_train = iris.data[indices_training] # keep for training all the matrix elements with the exception of the last 10
iris_y_train = iris.target[indices_training]
iris_X_test  = iris.data[indices_test] # keep the last 10 elements for test set
iris_y_test  = iris.target[indices_test]
```

# Fit the learning model on training set

```python
# fit the model to the training data
clf = clf.fit(iris_X_train, iris_y_train)
```

# Obtain predictions

```python
# apply fitted model "clf" to the test set
predicted_y_test = clf.predict(iris_X_test)

# print the predictions (class numbers associated to classes names in target names)
pred_y_test_labels = list(map(lambda val: iris.target_names[val], predicted_y_test))
iris_y_test_labels = list(map(lambda val: iris.target_names[val], iris_y_test))
print("Predictions:", pred_y_test_labels)
print("True classes:", iris_y_test_labels)
```

```python
list(np.array(pred_y_test_labels) == np.array(iris_y_test_labels))
```

Print the index of the test instances and the corresponding predictions

```python
# print the corresponding instances indexes and class names
for i in range(len(iris_y_test)):
    print("Instance # "+str(indices_test[i])+": ")
    print("Predicted: "+iris.target_names[predicted_y_test[i]]+"\t True: "+iris.target_names[iris_y_test[i]]+"\n")
```

# Look at the specific examples

```python
for i in range(len(iris_y_test)):
    print("Instance # "+str(indices_test[i])+": ")
    s=""
    for j in range(len(iris.feature_names)):
        s=s+iris.feature_names[j]+"="+str(iris_X_test[i][j])
        if (j<len(iris.feature_names)-1):
          s=s+", "
    print(s)
    print("Predicted: "+iris.target_names[predicted_y_test[i]]+"\t True: "+iris.target_names[iris_y_test[i]]+"\n")
```

# Obtain model performance results

```python
# print some metrics results
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
acc_score = accuracy_score(iris_y_test, predicted_y_test)
print("Accuracy score: "+ str(acc_score))
f1=f1_score(iris_y_test, predicted_y_test, average='macro')
print("F1 score: "+str(f1))
```

## TODO 3

Si apprendano gli alberi cercando di evitare l'overfitting (migliorando l'errore sul test set) facendo 'tuning' degli iper-parametri: il minimo numero dei campioni per foglia, la massima profondità dell'albero, i parametri di minomo decremento dell'impurezza, massimo numero dei nodi foglia, ecc. Si consiglia di usare la classe gridSearchCV per questo tipo di tuning di molti parametri.

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
# Generate a random permutation of the indices of examples that will be later used
# for the training and the test set
import numpy as np
iris_X_train, iris_X_test, iris_y_train, iris_y_test = train_test_split(iris.data, iris.target, test_size=0.1, random_state=42)
```

```python
from sklearn.model_selection import GridSearchCV

clf = tree.DecisionTreeClassifier(random_state=300, class_weight={0:1,1:1,2:1})

params = {
  "criterion": ["gini", "entropy", "log_loss"], # Function to measure the quality of a split.
  "max_depth": range(3, 8),                     # Maximum depth of the tree.
  "min_samples_leaf": range(2, 8),              # Minimum number of samples required to be at a leaf node.
  "max_leaf_nodes": range(3, 10),               # Grow a tree with max_leaf_nodes in best-first fashion.
  "min_samples_split": range(2, 10),            # Minimum number of samples required to split an internal node
  "min_impurity_decrease": [0 + i * 0.2 for i in range(int((2 - 0) / 0.2) + 1)],  # A node will be split if this split induces a decrease of the impurity greater than or equal to this value.
}

grid_search = GridSearchCV(clf, params)
grid_search.fit(iris_X_train, iris_y_train)
```

```python
grid_search.best_params_
```

```python
# apply fitted model "clf" to the test set
predicted_y_test = grid_search.predict(iris_X_test)

# Print f1 score and accuracy of the model
evaluate_model(iris_y_test, predicted_y_test)
```

## TODO 5

Si costruiscano le curve ROC (o curve nello spazio di coverage) e le si mostri per ciascun modello ad albero creato su un problema binario (con 1 sola classe positiva): per ciascun modello dovete costruire tre curve, una per ciascuna classe, considerata a turno la classe positiva.

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay
from sklearn import metrics
```

```python
def train_test_split_bin_clf(X, Y, label):

  label_indices = np.where(Y == label)[0]
  mask = np.full_like(Y, False)
  mask[label_indices] = True

  X_train, X_test, y_train, y_test = train_test_split(X, mask, test_size=0.2, random_state=42)

  return X_train, y_train, X_test, y_test
```

```python
import numpy as np

labels = np.unique(iris.target)
clf = tree.DecisionTreeClassifier(criterion="entropy",random_state=300,min_samples_leaf=5,class_weight={0:1,1:1,2:1})

plt.figure(0).clf()

for label in labels:
  X_train, y_train, X_test, y_test = train_test_split_bin_clf(iris.data, iris.target, label)

  # fit the model to the training data
  clf.fit(X_train, y_train)

  # apply fitted model "clf" to the test set
  y_pred = clf.predict(X_test)

  fpr, tpr, _ = metrics.roc_curve(y_test, y_pred)
  auc = round(metrics.roc_auc_score(y_test, y_pred), 4)
  plt.plot(fpr,tpr,label=f"{iris.target_names[label]}, AUC={auc}")

#add legend
plt.legend()
```
